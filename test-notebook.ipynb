```python
import yfinance as yf
import numpy as np

# Download 20 years of S&P 500 data
sp500 = yf.download('^GSPC', start='2004-01-01', end='2024-01-01')

# Calculate daily log returns (the primary observation variable for most financial HMMs)
sp500['Returns'] = np.log(sp500['Adj Close'] / sp500['Adj Close'].shift(1))
sp500 = sp500.dropna()
```
